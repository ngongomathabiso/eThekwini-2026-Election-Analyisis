```python
vd_totals = vd_totals.rename (
    columns = {"TotalValidVotes": "DistrictValidVotes"}
)
```

```python
party_vd = party_vd.merge (
    vd_totals,
    on = ["VotingDistrict", "BallotType"],
    how = "left"
)
```

```python
party_vd["VoteShare"] = (
    party_vd["TotalValidVotes"]
    / party_vd["DistrictValidVotes"]
) * 100
```
